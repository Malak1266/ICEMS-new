## 2. Chargement et Importation des Données

```python
# Chargement et préparation des données neurochirurgicales avec MÉTRIQUES ICEMS
print("📂 === CHARGEMENT DES DONNÉES ICEMS (FULL_DATA.JSON) ===")

class ICEMSSurgicalDataProcessor:
    """Processeur pour les données chirurgicales avec les 5 métriques ICEMS complètes"""
    
    def __init__(self, sequence_length=50):
        self.sequence_length = sequence_length
        self.scaler = StandardScaler()
        
        # Mapping des 9 niveaux d'expertise détaillés
        self.level_mapping = {
            'Medical student ': 0,
            'Resident PGY1': 1,
            'Resident PGY2': 2, 
            'Resident PGY3': 3,
            'Resident PGY4': 4,
            'Resident PGY5': 5,
            'Resident PGY6': 6,
            'Fellow': 7,
            'Fellow Pediatrics': 7,
            'Fellow Oncology ': 7,
            'Fellow functional': 7,
            'Fellow Epilepsy ': 7,
            'Fellow Spine': 7,
            'Fellow/Spine': 7,
            'Staff': 8
        }
        
        self.level_labels = [
            'Medical Student',    # 0
            'Resident PGY1',     # 1
            'Resident PGY2',     # 2
            'Resident PGY3',     # 3
            'Resident PGY4',     # 4
            'Resident PGY5',     # 5
            'Resident PGY6',     # 6
            'Fellow',            # 7
            'Staff'              # 8
        ]
        
        # Scores continus pour régression (0-1)
        self.continuous_mapping = {
            0: 0.0,      # Medical Student
            1: 0.125,    # PGY1
            2: 0.25,     # PGY2
            3: 0.375,    # PGY3
            4: 0.5,      # PGY4
            5: 0.625,    # PGY5
            6: 0.75,     # PGY6
            7: 0.875,    # Fellow
            8: 1.0       # Staff
        }
    
    def load_icems_data(self, data_path='data/full_data.json'):
        """Charge les données ICEMS avec la structure réelle identifiée"""
        print(f"📂 Chargement des données ICEMS depuis {data_path}")
        
        if not os.path.exists(data_path):
            raise FileNotFoundError(f"Fichier non trouvé: {data_path}")
        
        try:
            with open(data_path, 'r', encoding='utf-8') as f:
                raw_data = json.load(f)
            
            print(f"✅ Données ICEMS chargées: {len(raw_data)} sections")
            return self._process_icems_data_correct(raw_data)
            
        except Exception as e:
            print(f"❌ Erreur lors du chargement: {e}")
            raise
    
    def _process_icems_data_correct(self, raw_data):
        """Traitement correct basé sur la structure analysée"""
        all_sequences = []
        all_labels_discrete = []
        all_labels_continuous = []
        
        print("🔧 Traitement des données ICEMS avec structure correcte...")
        
        # Extraire les sections nécessaires
        participants = raw_data['participant']
        trials = raw_data['trial']
        levels = raw_data['level']
        instruments = raw_data['instrument']
        metrics = raw_data['metric']
        data_arrays = raw_data['data']
        
        print(f"📊 Structure confirmée: {len(data_arrays)} séries de données")
        
        # Organiser les indices par métriques
        metric_indices = {}
        for idx, metric_name in metrics.items():
            if metric_name not in metric_indices:
                metric_indices[metric_name] = []
            metric_indices[metric_name].append(idx)
        
        print(f"📊 Métriques organisées:")
        for metric, indices in metric_indices.items():
            print(f"   {metric}: {len(indices)} séries")
        
        # Organiser par participant pour créer des échantillons cohérents
        participant_data = {}
        for idx in data_arrays.keys():
            participant_id = participants[idx]
            trial_id = trials[idx]
            level_name = levels[idx]
            instrument_name = instruments[idx]
            metric_name = metrics[idx]
            
            # Créer une clé unique pour chaque échantillon
            sample_key = f"{participant_id}_{trial_id}_{instrument_name}"
            
            if sample_key not in participant_data:
                participant_data[sample_key] = {
                    'level': level_name,
                    'participant': participant_id,
                    'trial': trial_id,
                    'instrument': instrument_name,
                    'metrics': {}
                }
            
            participant_data[sample_key]['metrics'][metric_name] = {
                'index': idx,
                'data': data_arrays[idx]
            }
        
        print(f"📊 Échantillons organisés: {len(participant_data)} échantillons uniques")
        
        # Traiter chaque échantillon
        processed_count = 0
        skipped_count = 0
        
        for sample_key, sample_info in participant_data.items():
            level_name = sample_info['level']
            
            if level_name not in self.level_mapping:
                skipped_count += 1
                continue
            
            try:
                # Extraire les métriques ICEMS pour cet échantillon
                icems_features = self._extract_icems_for_sample(sample_info['metrics'])
                
                if icems_features is None or icems_features.shape[0] < self.sequence_length:
                    skipped_count += 1
                    continue
                
                # Mapper le niveau d'expertise
                level_id = self.level_mapping[level_name]
                
                # Créer les séquences temporelles
                sequences = self._create_sequences(icems_features, self.sequence_length)
                
                # Ajouter les séquences et labels
                for sequence in sequences:
                    all_sequences.append(sequence)
                    all_labels_discrete.append(level_id)
                    all_labels_continuous.append(self.continuous_mapping[level_id])
                
                processed_count += 1
                
                # Debug pour les premiers échantillons
                if processed_count <= 5:
                    print(f"   ✅ {sample_key}: {level_name} → {icems_features.shape} → {len(sequences)} séquences")
                    
            except Exception as e:
                skipped_count += 1
                if processed_count <= 3:
                    print(f"   ❌ Erreur {sample_key}: {str(e)[:100]}")
                continue
        
        print(f"📊 Traitement terminé: {processed_count} traités, {skipped_count} ignorés")
        
        if len(all_sequences) == 0:
            raise ValueError("Aucune séquence ICEMS valide extraite")
        
        X = np.array(all_sequences)
        y_discrete = np.array(all_labels_discrete)
        y_continuous = np.array(all_labels_continuous)
        
        # Distribution finale
        unique_levels, counts = np.unique(y_discrete, return_counts=True)
        final_distribution = {}
        for level_id, count in zip(unique_levels, counts):
            level_name = self.level_labels[level_id]
            final_distribution[level_name] = count
        
        print(f"✅ Extraction ICEMS terminée: {len(all_sequences)} séquences")
        print(f"📊 Distribution finale des niveaux:")
        for level_name, count in final_distribution.items():
            percentage = (count / len(all_sequences)) * 100
            print(f"   {level_name}: {count} séquences ({percentage:.1f}%)")
        
        return X, y_discrete, y_continuous, final_distribution
    
    def _extract_icems_for_sample(self, sample_metrics):
        """Extrait les 5 métriques ICEMS avec filtrage pour un échantillon"""
        try:
            # Dictionnaire pour stocker les métriques
            extracted_metrics = {}
            captured_flags = {}
            
            # Extraire les métriques de base et leurs flags
            for metric_name, metric_info in sample_metrics.items():
                data_array = np.array(metric_info['data'])
                
                if metric_name in ['velocity', 'acceleration', 'jerk']:
                    extracted_metrics[metric_name] = data_array
                elif metric_name == 'captured_flag':
                    captured_flags['individual'] = data_array.astype(bool)
                elif metric_name == 'distance':
                    if 'distance_1' not in extracted_metrics:
                        extracted_metrics['distance_1'] = data_array
                    else:
                        extracted_metrics['distance_2'] = data_array
                elif metric_name == 'bidist_captured_flag':
                    captured_flags['bimanual'] = data_array.astype(bool)
            
            # Vérifier qu'on a les métriques essentielles
            required_metrics = ['velocity', 'acceleration', 'jerk']
            if not all(metric in extracted_metrics for metric in required_metrics):
                return None
            
            # Appliquer le filtrage avec captured_flag pour les métriques individuelles
            filtered_features = []
            
            if 'individual' in captured_flags:
                captured_flag = captured_flags['individual']
                
                for metric in required_metrics:
                    if metric in extracted_metrics:
                        metric_data = extracted_metrics[metric]
                        if len(metric_data) == len(captured_flag):
                            # Filtrer avec captured_flag
                            filtered_data = metric_data[captured_flag]
                            if len(filtered_data) > 0:
                                filtered_features.append(filtered_data)
            else:
                # Si pas de captured_flag, utiliser les données brutes
                for metric in required_metrics:
                    if metric in extracted_metrics:
                        filtered_features.append(extracted_metrics[metric])
            
            # Traiter les distances avec bidist_captured_flag
            if 'bimanual' in captured_flags and any('distance' in k for k in extracted_metrics.keys()):
                bidist_flag = captured_flags['bimanual']
                
                for dist_key in ['distance_1', 'distance_2']:
                    if dist_key in extracted_metrics:
                        distance_data = extracted_metrics[dist_key]
                        if len(distance_data) == len(bidist_flag):
                            filtered_distance = distance_data[bidist_flag]
                            if len(filtered_distance) > 0:
                                # Interpoler pour matcher la longueur des autres métriques
                                if filtered_features and len(filtered_distance) != len(filtered_features[0]):
                                    if len(filtered_distance) > 1 and len(filtered_features[0]) > 1:
                                        x_old = np.linspace(0, 1, len(filtered_distance))
                                        x_new = np.linspace(0, 1, len(filtered_features[0]))
                                        filtered_distance = np.interp(x_new, x_old, filtered_distance)
                                    elif len(filtered_distance) == 1:
                                        filtered_distance = np.full(len(filtered_features[0]), filtered_distance[0])
                                
                                filtered_features.append(filtered_distance)
            
            if len(filtered_features) == 0:
                return None
            
            # Assurer que toutes les métriques ont la même longueur
            min_length = min(len(feature) for feature in filtered_features)
            if min_length < self.sequence_length:
                return None
            
            # Tronquer et créer la matrice finale
            features_matrix = np.array([feature[:min_length] for feature in filtered_features])
            
            # Transposer pour avoir (temps, features)
            features_matrix = features_matrix.T
            
            return features_matrix
            
        except Exception as e:
            return None
    
    def _create_sequences(self, data, seq_length):
        """Crée des séquences temporelles NON-chevauchantes à partir des données ICEMS"""
        sequences = []
        
        if len(data) < seq_length:
            # Padding si trop court
            padded_data = np.zeros((seq_length, data.shape[1]))
            padded_data[:len(data)] = data
            sequences.append(padded_data)
        else:
            # Séquences NON-chevauchantes (step_size = seq_length)
            step_size = seq_length  # ⚡ MODIFIÉ: Pas de chevauchement
            for i in range(0, len(data) - seq_length + 1, step_size):
                sequences.append(data[i:i + seq_length])
        
        return sequences

    def preprocess_data(self, X, y_discrete, y_continuous):
        """Prétraitement des données ICEMS avec normalisation Z-score"""
        print("🔧 Prétraitement des données ICEMS...")
        
        # Normalisation Z-score par feature
        original_shape = X.shape
        X_reshaped = X.reshape(-1, X.shape[-1])
        X_normalized = self.scaler.fit_transform(X_reshaped)
        X = X_normalized.reshape(original_shape)
        
        print(f"✅ Données ICEMS normalisées:")
        print(f"   Forme: {X.shape}")
        print(f"   {X.shape[-1]} métriques ICEMS par pas de temps")
        print(f"   Moyenne: {X.mean():.6f}")
        print(f"   Écart-type: {X.std():.6f}")
        print(f"   Plage: [{X.min():.3f}, {X.max():.3f}]")
        
        return X, y_discrete, y_continuous

# 🚀 CHARGEMENT DES DONNÉES ICEMS COMPLÈTES
print("🧠 === INITIALISATION DU PROCESSEUR ICEMS ===")
processor = ICEMSSurgicalDataProcessor(sequence_length=50)
print("✅ Processeur ICEMS initialisé pour 9 niveaux d'expertise")

# Chargement des données ICEMS
print("\n📂 === CHARGEMENT DES DONNÉES ICEMS (FULL_DATA.JSON) ===")
try:
    X, y_discrete, y_continuous, level_distribution = processor.load_icems_data()
    print(f"✅ Données ICEMS chargées avec succès")
except Exception as e:
    print(f"❌ ERREUR: {e}")
    print("🔧 Problème avec l'extraction des métriques ICEMS")
    raise

# Prétraitement ICEMS
print("\n🔧 === PRÉTRAITEMENT ICEMS ===")
X, y_discrete, y_continuous = processor.preprocess_data(X, y_discrete, y_continuous)

print("\n🎉 === DONNÉES ICEMS PRÊTES ===")
print(f"📊 Forme finale des données:")
print(f"   X (séquences): {X.shape}")
print(f"   y_discrete (niveaux): {y_discrete.shape} - {len(np.unique(y_discrete))} classes uniques")  
print(f"   y_continuous (scores): {y_continuous.shape} - plage [{y_continuous.min():.3f}, {y_continuous.max():.3f}]")

# Affichage final de la distribution
print(f"\n🎯 === DISTRIBUTION FINALE DES NIVEAUX ICEMS ===")
unique_levels, counts = np.unique(y_discrete, return_counts=True)
total_sequences = len(y_discrete)

for level_id, count in zip(unique_levels, counts):
    if level_id < len(processor.level_labels):
        level_name = processor.level_labels[level_id]
        percentage = (count / total_sequences) * 100
        continuous_score = processor.continuous_mapping[level_id]
        print(f"   {level_id}: {level_name} → {count} séquences ({percentage:.1f}%) [score: {continuous_score:.3f}]")

print(f"\n✅ === DONNÉES ICEMS PRÊTES POUR L'ENTRAÎNEMENT ===")
print(f"📈 {total_sequences} séquences de {X.shape[1]} pas de temps")
print(f"🎯 {X.shape[2]} métriques ICEMS extraites et filtrées")
print(f"🔬 {len(np.unique(y_discrete))} niveaux d'expertise détectés")
print(f"⚙️ Filtrage ICEMS appliqué: captured_flag + bidist_captured_flag")
```

```python
# Architecture hybride ultra-avancée corrigée
print("🏗️ === ARCHITECTURE HYBRIDE ULTRA-AVANCÉE CORRIGÉE ===")

def create_positional_encoding(max_len, d_model):
    """Crée un encodage positionnel pour le Transformer"""
    pos_encoding = np.zeros((max_len, d_model))
    
    for pos in range(max_len):
        for i in range(0, d_model, 2):
            pos_encoding[pos, i] = np.sin(pos / (10000 ** ((2 * i) / d_model)))
            if i + 1 < d_model:
                pos_encoding[pos, i + 1] = np.cos(pos / (10000 ** ((2 * (i + 1)) / d_model)))
    
    return tf.constant(pos_encoding, dtype=tf.float32)

def multi_head_attention_block(inputs, num_heads, key_dim, dropout_rate=0.1, name_prefix=""):
    """Bloc d'attention multi-têtes optimisé"""
    
    # Attention multi-têtes
    attention = MultiHeadAttention(
        num_heads=num_heads,
        key_dim=key_dim,
        dropout=dropout_rate,
        name=f'{name_prefix}_mha'
    )(inputs, inputs)
    
    # Connexion résiduelle + normalisation
    attention = Add(name=f'{name_prefix}_add1')([inputs, attention])
    attention = LayerNormalization(name=f'{name_prefix}_ln1')(attention)
    
    # Feed-forward network
    ffn = Dense(key_dim * 4, activation='relu', name=f'{name_prefix}_ffn1')(attention)
    ffn = Dropout(dropout_rate, name=f'{name_prefix}_ffn_dropout')(ffn)
    ffn = Dense(inputs.shape[-1], name=f'{name_prefix}_ffn2')(ffn)
    
    # Connexion résiduelle + normalisation
    output = Add(name=f'{name_prefix}_add2')([attention, ffn])
    output = LayerNormalization(name=f'{name_prefix}_ln2')(output)
    
    return output

def create_hybrid_advanced_model(input_shape, num_classes=9):
    """
    🧠 Architecture Hybride Ultra-Avancée pour l'Expertise Chirurgicale
    Version corrigée sans les fonctions problématiques
    """
    
    inputs = Input(shape=input_shape, name='surgical_sequence')
    
    # === SECTION 1: PREPROCESSING & EMBEDDINGS AVANCÉS ===
    
    # Projection vers dimensions d'attention optimales
    d_model = 256  # Réduit pour éviter les problèmes de mémoire
    embedded = Dense(d_model, activation='relu', name='feature_projection')(inputs)
    embedded = LayerNormalization(name='input_norm')(embedded)
    embedded = Dropout(0.1, name='input_dropout')(embedded)
    
    # Encodage positionnel simplifié
    seq_len = input_shape[0]
    pos_encoding = create_positional_encoding(seq_len, d_model)
    
    # Ajouter l'encodage positionnel de manière compatible
    embedded_with_pos = Lambda(
        lambda x: x + pos_encoding[:tf.shape(x)[1], :], 
        name='positional_encoding'
    )(embedded)
    
    # === SECTION 2: TRANSFORMER ENCODER STACK ===
    
    transformer_output = embedded_with_pos
    
    # Premier bloc Transformer
    transformer_output = multi_head_attention_block(
        transformer_output, num_heads=8, key_dim=32, dropout_rate=0.1, name_prefix="transformer_1"
    )
    
    # Deuxième bloc Transformer
    transformer_output = multi_head_attention_block(
        transformer_output, num_heads=6, key_dim=32, dropout_rate=0.1, name_prefix="transformer_2"
    )
    
    # === SECTION 3: LSTM PARALLEL PROCESSING AVANCÉE ===
    
    # Première branche LSTM bidirectionnelle
    lstm_branch1 = Bidirectional(
        LSTM(128, return_sequences=True, dropout=0.2, recurrent_dropout=0.1),
        name='lstm_primary'
    )(embedded_with_pos)
    lstm_branch1 = BatchNormalization(name='lstm_bn1')(lstm_branch1)
    
    # Deuxième branche LSTM
    lstm_branch2 = Bidirectional(
        LSTM(96, return_sequences=True, dropout=0.2, recurrent_dropout=0.1),
        name='lstm_secondary'  
    )(lstm_branch1)
    lstm_branch2 = BatchNormalization(name='lstm_bn2')(lstm_branch2)
    
    # === SECTION 4: GRU SPÉCIALISÉ POUR PATTERNS CHIRURGICAUX ===
    
    gru_branch1 = Bidirectional(
        GRU(96, return_sequences=True, dropout=0.2),
        name='gru_primary'
    )(embedded_with_pos)
    gru_branch1 = BatchNormalization(name='gru_bn1')(gru_branch1)
    
    gru_branch2 = Bidirectional(
        GRU(64, return_sequences=True, dropout=0.2),
        name='gru_secondary'
    )(gru_branch1)
    gru_branch2 = BatchNormalization(name='gru_bn2')(gru_branch2)
    
    # === SECTION 5: CNN POUR PATTERNS LOCAUX ===
    
    cnn_branch = Conv1D(128, 3, padding='same', activation='relu', name='cnn1')(embedded_with_pos)
    cnn_branch = BatchNormalization(name='cnn_bn1')(cnn_branch)
    cnn_branch = Dropout(0.1, name='cnn_dropout1')(cnn_branch)
    
    cnn_branch = Conv1D(256, 3, padding='same', activation='relu', name='cnn2')(cnn_branch)
    cnn_branch = BatchNormalization(name='cnn_bn2')(cnn_branch)
    
    # === SECTION 6: CROSS-ATTENTION FUSION AVANCÉE ===
    
    # Fusion des représentations
    all_features = [transformer_output, lstm_branch2, gru_branch2, cnn_branch]
    fused = Concatenate(name='feature_fusion')(all_features)
    
    # Cross-attention entre modalités
    cross_attention = MultiHeadAttention(
        num_heads=6, key_dim=32, dropout=0.1, name='cross_attention'
    )(fused, fused)
    
    fused = Add(name='cross_add')([fused, cross_attention])
    fused = LayerNormalization(name='cross_ln')(fused)
    
    # === SECTION 7: MULTI-SCALE POOLING SIMPLIFIÉ ===
    
    # Pooling global standard
    global_max = GlobalMaxPooling1D(name='global_max')(fused)
    global_avg = GlobalAveragePooling1D(name='global_avg')(fused)
    
    # Attention temporelle pondérée
    temporal_attention = Dense(fused.shape[-1], activation='softmax', name='temporal_attention_weights')(fused)
    weighted_features = Multiply(name='temporal_weighting')([fused, temporal_attention])
    global_weighted = GlobalAveragePooling1D(name='global_weighted')(weighted_features)
    
    # Last timestep (important pour les séquences)
    last_timestep = Lambda(lambda x: x[:, -1, :], name='last_timestep')(fused)
    
    # Pooling multi-fenêtres simplifié
    # Diviser en segments et faire la moyenne
    segment_size = input_shape[0] // 5  # 5 segments
    segment_pools = []
    
    for i in range(5):
        start_idx = i * segment_size
        end_idx = min((i + 1) * segment_size, input_shape[0])
        segment = Lambda(
            lambda x, s=start_idx, e=end_idx: x[:, s:e, :], 
            name=f'segment_{i}'
        )(fused)
        segment_pool = GlobalAveragePooling1D(name=f'segment_pool_{i}')(segment)
        segment_pools.append(segment_pool)
    
    # Combiner les segments
    multi_segment_pool = Concatenate(name='multi_segment')(segment_pools)
    
    # === SECTION 8: FEATURE INTEGRATION AVANCÉE ===
    
    # Combinaison de toutes les représentations
    combined_features = Concatenate(name='multi_scale_features')([
        global_max, global_avg, global_weighted, last_timestep, multi_segment_pool
    ])
    
    # === SECTION 9: CLASSIFICATION HEAD SIMPLIFIÉE ===
    
    # Branche principale avec connexions résiduelles
    main_branch = Dense(512, activation='relu', 
                       kernel_regularizer=l1_l2(l1=1e-5, l2=1e-4),
                       name='main_dense_1')(combined_features)
    main_branch = BatchNormalization(name='main_bn_1')(main_branch)
    main_branch = Dropout(0.4, name='main_dropout_1')(main_branch)
    
    main_branch = Dense(256, activation='relu', 
                       kernel_regularizer=l1_l2(l1=1e-5, l2=1e-4),
                       name='main_dense_2')(main_branch)
    main_branch = BatchNormalization(name='main_bn_2')(main_branch)
    main_branch = Dropout(0.3, name='main_dropout_2')(main_branch)
    
    main_branch = Dense(128, activation='relu', name='main_dense_3')(main_branch)
    main_branch = BatchNormalization(name='main_bn_3')(main_branch)
    main_branch = Dropout(0.2, name='main_dropout_3')(main_branch)
    
    # === SORTIE UNIQUE POUR ÉVITER LES COMPLICATIONS ===
    
    # Sortie classification (9 classes)
    classification_output = Dense(num_classes, activation='softmax', 
                                 name='classification_output')(main_branch)
    
    # === CRÉATION DU MODÈLE SIMPLIFIÉ ===
    
    model = Model(
        inputs=inputs, 
        outputs=classification_output,
        name='HybridSurgicalExpertisePredictor'
    )
    
    return model

def create_focal_loss(alpha=0.25, gamma=2.0):
    """Fonction de perte Focal Loss pour les classes déséquilibrées"""
    
    def focal_loss(y_true, y_pred):
        # Éviter log(0)
        y_pred = tf.clip_by_value(y_pred, tf.keras.backend.epsilon(), 1.0 - tf.keras.backend.epsilon())
        
        # Calculer la cross-entropy
        ce_loss = tf.keras.losses.sparse_categorical_crossentropy(y_true, y_pred)
        
        # Calculer le facteur focal
        pt = tf.exp(-ce_loss)
        focal_loss_value = alpha * (1 - pt) ** gamma * ce_loss
        
        return tf.reduce_mean(focal_loss_value)
    
    return focal_loss

# === CONSTRUCTION DU MODÈLE HYBRIDE CORRIGÉ ===

print("🏗️ Création du modèle hybride ultra-avancé corrigé...")

input_shape = X_train.shape[1:]
hybrid_model = create_hybrid_advanced_model(input_shape, num_classes=9)

# Compilation avec fonction de perte Focal Loss
hybrid_model.compile(
    optimizer=Adam(learning_rate=0.0008, beta_1=0.9, beta_2=0.999, epsilon=1e-7),
    loss=create_focal_loss(alpha=0.25, gamma=2.0),
    metrics=['accuracy']
)

print("✅ Modèle hybride ultra-avancé créé avec succès!")
print(f"📊 Paramètres totaux: {hybrid_model.count_params():,}")
print(f"💾 Taille estimée: ~{hybrid_model.count_params() * 4 / 1e6:.1f} MB")

print(f"\n🚀 === INNOVATIONS INTÉGRÉES (VERSION CORRIGÉE) ===")
print("✅ Architecture Transformer + LSTM + GRU + CNN")
print("✅ Double niveau d'attention multi-têtes")
print("✅ Encodage positionnel optimisé")
print("✅ Cross-attention entre modalités")
print("✅ Multi-scale pooling avec segments temporels")
print("✅ Fonction de perte Focal Loss pour déséquilibre")
print("✅ Connexions résiduelles adaptatives")
print("✅ Architecture simplifiée pour éviter les erreurs")

print(f"\n🔧 === CORRECTIONS APPORTÉES ===")
print("🔧 Suppression du dual-output pour simplifier")
print("🔧 Remplacement du windowed pooling par segments fixes")
print("🔧 Réduction des dimensions (384→256) pour stabilité")
print("🔧 Focal Loss simple au lieu de loss hybride complexe")
print("🔧 Architecture plus robuste et compatible")

# Affichage de l'architecture
hybrid_model.summary()
```

```python
# Entraînement ultra-optimisé corrigé pour single-output
print("🚀 === ENTRAÎNEMENT HYBRIDE ULTRA-OPTIMISÉ CORRIGÉ ===")

def advanced_data_augmentation(X, y_disc, y_cont, augmentation_factor=0.2):
    """Augmentation de données spécialisée pour chirurgie"""
    print(f"🔄 Augmentation avancée (facteur: {augmentation_factor})")
    
    n_aug = int(len(X) * augmentation_factor)
    indices = np.random.choice(len(X), n_aug, replace=True)
    
    X_aug = X[indices].copy()
    y_disc_aug = y_disc[indices].copy()
    y_cont_aug = y_cont[indices].copy()
    
    # Techniques d'augmentation chirurgicale
    for i in range(len(X_aug)):
        expertise_level = y_cont_aug[i]
        
        # 1. Bruit adaptatif basé sur l'expertise
        noise_std = 0.01 if expertise_level > 0.7 else 0.03
        noise = np.random.normal(0, noise_std, X_aug[i].shape)
        X_aug[i] += noise
        
        # 2. Décalage temporel chirurgical
        if np.random.random() > 0.6:
            shift = np.random.randint(-3, 4)
            if shift != 0:
                X_aug[i] = np.roll(X_aug[i], shift, axis=0)
        
        # 3. Mise à l'échelle basée sur la variabilité expertise
        if np.random.random() > 0.7:
            scale_range = 0.05 if expertise_level > 0.5 else 0.1
            scale_factor = np.random.uniform(1-scale_range, 1+scale_range)
            X_aug[i] *= scale_factor
        
        # 4. Masquage temporel aléatoire
        if np.random.random() > 0.8:
            mask_length = np.random.randint(1, 4)
            mask_start = np.random.randint(0, X_aug[i].shape[0] - mask_length)
            X_aug[i][mask_start:mask_start+mask_length] *= 0.1
    
    return (np.concatenate([X, X_aug]), 
            np.concatenate([y_disc, y_disc_aug]),
            np.concatenate([y_cont, y_cont_aug]))

def create_advanced_callbacks():
    """Callbacks ultra-optimisés pour single-output"""
    
    return [
        EarlyStopping(
            monitor='val_accuracy',  # Corrigé pour single-output
            patience=20,
            restore_best_weights=True,
            verbose=1,
            min_delta=0.001
        ),
        ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.8,
            patience=12,
            min_lr=1e-8,
            verbose=1,
            cooldown=5
        ),
        ModelCheckpoint(
            'model_outputs/hybrid_surgical_best.keras',
            monitor='val_accuracy',  # Corrigé pour single-output
            save_best_only=True,
            verbose=1
        ),
        LearningRateScheduler(
            lambda epoch: 0.0008 * (0.96 ** epoch) if epoch < 40 else 0.0001 * (0.98 ** (epoch-40))
        )
    ]

# === CORRECTION DES VARIABLES POUR ÉVITER L'ERREUR DE CARDINALITÉ ===
print("\n🔧 === CORRECTION DES VARIABLES D'ENTRAÎNEMENT ===")

# Utiliser les bonnes variables (sans _scaled pour les données d'entraînement)
print(f"📊 Variables disponibles:")
print(f"   X_train: {X_train.shape}")
print(f"   X_val: {X_val.shape}")
print(f"   X_test: {X_test.shape}")
print(f"   y_train_disc: {y_train_disc.shape}")
print(f"   y_val_disc: {y_val_disc.shape}")

# Correction des variables pour éviter la confusion
X_train_scaled = X_train  # Les données sont déjà normalisées
X_val_scaled = X_val
X_test_scaled = X_test

print(f"✅ Variables corrigées:")
print(f"   X_train_scaled: {X_train_scaled.shape}")
print(f"   X_val_scaled: {X_val_scaled.shape}")
print(f"   X_test_scaled: {X_test_scaled.shape}")

# === PRÉPARATION DES DONNÉES POUR SINGLE-OUTPUT ===
print("\n📊 Préparation des données pour classification single-output...")

# Augmentation des données d'entraînement avec les bonnes variables
X_train_aug, y_train_disc_aug, y_train_cont_aug = advanced_data_augmentation(
    X_train_scaled, y_train_disc, y_train_cont, augmentation_factor=0.25
)

print(f"✅ Données augmentées: {len(X_train_scaled):,} → {len(X_train_aug):,}")

# Vérification de la cohérence des données
print(f"\n🔍 === VÉRIFICATION DE COHÉRENCE ===")
print(f"   X_train_aug: {X_train_aug.shape}")
print(f"   y_train_disc_aug: {y_train_disc_aug.shape}")
print(f"   X_val_scaled: {X_val_scaled.shape}")
print(f"   y_val_disc: {y_val_disc.shape}")

# === CONFIGURATION D'ENTRAÎNEMENT CORRIGÉE ===
print("\n⚙️ Configuration d'entraînement ultra-optimisée...")

callbacks = create_advanced_callbacks()

# Configuration pour single-output (classification uniquement)
train_config = {
    'epochs': 100,
    'batch_size': 12,  # Batch petit pour précision maximale
    'validation_data': (X_val_scaled, y_val_disc),  # Données de validation cohérentes
    'callbacks': callbacks,
    'sample_weight': np.array([class_weight_dict[y] for y in y_train_disc_aug]),  # Poids corrects
    'verbose': 1,
    'shuffle': True
}

print(f"✅ Config: {train_config['epochs']} epochs, batch_size={train_config['batch_size']}")
print(f"⚖️ Poids de classe appliqués pour équilibrage")

# Vérification finale avant l'entraînement
print(f"\n🔍 === VÉRIFICATION FINALE AVANT ENTRAÎNEMENT ===")
print(f"   Données d'entraînement X: {X_train_aug.shape}")
print(f"   Labels d'entraînement y: {y_train_disc_aug.shape}")
print(f"   Sample weights: {len([class_weight_dict[y] for y in y_train_disc_aug])}")
print(f"   Données de validation X: {X_val_scaled.shape}")
print(f"   Labels de validation y: {y_val_disc.shape}")

# === ENTRAÎNEMENT PRINCIPAL CORRIGÉ ===
print(f"\n🏋️ === DÉMARRAGE ENTRAÎNEMENT CLASSIFICATION ===")
print(f"📊 Données: {X_train_aug.shape}")
print(f"🎯 Objectif: Classification (9 classes)")

start_time = time.time()

try:
    # Entraînement avec single-output et données cohérentes
    history = hybrid_model.fit(
        X_train_aug, y_train_disc_aug,  # Données d'entraînement cohérentes
        **train_config
    )
    
    training_time = time.time() - start_time
    print(f"\n⏱️ Entraînement terminé en {training_time:.1f} secondes")
    
    # === ÉVALUATION POST-ENTRAÎNEMENT ===
    print(f"\n📊 === ÉVALUATION POST-ENTRAÎNEMENT ===")
    
    # Prédictions classification
    train_preds = hybrid_model.predict(X_train_aug, verbose=0)
    val_preds = hybrid_model.predict(X_val_scaled, verbose=0)
    
    # Métriques classification
    train_acc = accuracy_score(y_train_disc_aug, np.argmax(train_preds, axis=1))
    val_acc = accuracy_score(y_val_disc, np.argmax(val_preds, axis=1))
    
    # Métriques régression simulées sur les sorties continues
    train_pred_cont = np.argmax(train_preds, axis=1) / 8.0  # Convertir en score continu
    val_pred_cont = np.argmax(val_preds, axis=1) / 8.0
    
    train_r2 = r2_score(y_train_cont_aug, train_pred_cont)
    val_r2 = r2_score(y_val_cont, val_pred_cont)
    
    train_mae = mean_absolute_error(y_train_cont_aug, train_pred_cont)
    val_mae = mean_absolute_error(y_val_cont, val_pred_cont)
    
    print(f"\n🏆 === RÉSULTATS FINAUX ===")
    print("=" * 60)
    print(f"{'Métrique':<25} {'Train':<15} {'Validation':<15}")
    print("-" * 60)
    print(f"{'Classification Accuracy':<25} {train_acc:<15.4f} {val_acc:<15.4f}")
    print(f"{'Regression R² (simulé)':<25} {train_r2:<15.4f} {val_r2:<15.4f}")
    print(f"{'Regression MAE (simulé)':<25} {train_mae:<15.4f} {val_mae:<15.4f}")
    print("=" * 60)
    
    # Diagnostic de surapprentissage
    acc_gap = train_acc - val_acc
    r2_gap = train_r2 - val_r2
    
    print(f"\n🔍 === DIAGNOSTIC SURAPPRENTISSAGE ===")
    print(f"📊 Gap Accuracy: {acc_gap:.4f}")
    print(f"📊 Gap R² (simulé): {r2_gap:.4f}")
    
    if acc_gap < 0.1 and r2_gap < 0.15:
        print("✅ Excellent équilibre - Pas de surapprentissage")
    elif acc_gap < 0.2 and r2_gap < 0.25:
        print("🟡 Léger surapprentissage - Acceptable")
    else:
        print("🔴 Surapprentissage détecté - Augmenter régularisation")
    
    # Sauvegarde des résultats
    results = {
        'model': hybrid_model,
        'history': history,
        'metrics': {
            'train_accuracy': train_acc,
            'val_accuracy': val_acc,
            'train_r2': train_r2,
            'val_r2': val_r2,
            'train_mae': train_mae,
            'val_mae': val_mae
        },
        'training_time': training_time,
        'predictions': {
            'val_classification': val_preds,
            'val_regression': val_pred_cont
        }
    }
    
    print(f"\n✅ === ENTRAÎNEMENT HYBRIDE TERMINÉ AVEC SUCCÈS ===")
    print(f"🎯 Accuracy: {val_acc:.4f} ({val_acc*100:.1f}%)")
    print(f"📊 R² simulé: {val_r2:.4f}")
    print(f"💾 Modèle sauvegardé: model_outputs/hybrid_surgical_best.keras")
    
except Exception as e:
    print(f"❌ Erreur durant l'entraînement: {e}")
    import traceback
    traceback.print_exc()
    results = None

print(f"\n🚀 === MODÈLE HYBRIDE ULTRA-AVANCÉ PRÊT ===")
print(f"🎯 Performance attendue: Classification >75%")
print(f"💡 Architecture: {hybrid_model.count_params():,} paramètres")
print(f"🔧 Mode: Single-output classification optimisé")
```

```python
# === ANALYSE DES POIDS DE CONTRIBUTION DES MÉTRIQUES ===

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.inspection import permutation_importance
import pandas as pd
import json 
print("🔍 === ANALYSE DES POIDS DE CONTRIBUTION DES MÉTRIQUES ===")
print("📊 Calcul de l'importance de chaque feature biomécanique...")

# Définition des noms des métriques biomécaniques (VRAIES MÉTRIQUES DU DATASET)
feature_names = [
    'velocity', 'acceleration', 'jerk',
    'jerk_x', 'jerk_y', 'jerk_z',
    'efficiency_index', 'average_separation_distance'
]

# Remarque: timestamp n'est pas inclus car c'est une dimension temporelle, pas une feature

print(f"✅ Nombre de features: {len(feature_names)}")

try:
    # === MÉTHODE 1: IMPORTANCE PAR PERMUTATION ===
    print("\n🔄 Calcul de l'importance par permutation...")
    
    # Utilisation du modèle final entraîné
    model_to_analyze = adaptive_model if 'adaptive_model' in globals() else hybrid_model
    
    # Reshaping des données pour l'analyse d'importance
    # Flatten les données temporelles pour l'analyse de permutation
    X_val_flat = X_val_scaled.reshape(X_val_scaled.shape[0], -1)
    
    # Sélection d'un sous-échantillon pour accélérer le calcul
    n_samples = min(100, len(X_val_scaled))  # Limiter pour performance
    sample_indices = np.random.choice(len(X_val_scaled), n_samples, replace=False)
    
    X_sample = X_val_scaled[sample_indices]
    y_sample = y_val_cont[sample_indices]
    
    print(f"📊 Échantillon d'analyse: {n_samples} exemples")
    
    # === MÉTHODE 2: ANALYSE DES POIDS DU MODÈLE ===
    print("\n⚖️ Analyse des poids des couches du modèle...")
    
    # Extraction des poids de la première couche dense significative
    layer_weights = None
    layer_name = "Aucune couche trouvée"
    
    for i, layer in enumerate(model_to_analyze.layers):
        if hasattr(layer, 'get_weights') and len(layer.get_weights()) > 0:
            weights = layer.get_weights()[0]
            if len(weights.shape) >= 2 and weights.shape[0] > 1:
                layer_weights = weights
                layer_name = f"{layer.name} (couche {i+1})"
                break
    
    # === MÉTHODE 3: ANALYSE STATISTIQUE DES CORRELATIONS ===
    print("\n📈 Analyse des corrélations feature-target...")
    
    # Calcul des corrélations moyennes par feature
    correlations = []
    feature_importance_stats = []
    
    for feature_idx in range(len(feature_names)):
        # Extraction de la feature à travers tous les timesteps
        feature_values = X_val_scaled[:, :, feature_idx].flatten()
        target_values = np.repeat(y_val_cont, X_val_scaled.shape[1])
        
        # Corrélation Pearson
        correlation = np.corrcoef(feature_values, target_values)[0, 1]
        correlations.append(abs(correlation) if not np.isnan(correlation) else 0)
        
        # Statistiques descriptives
        feature_stats = {
            'name': feature_names[feature_idx],
            'correlation': abs(correlation) if not np.isnan(correlation) else 0,
            'std': np.std(feature_values),
            'range': np.ptp(feature_values),  # peak-to-peak (range)
            'mean_abs': np.mean(np.abs(feature_values))
        }
        feature_importance_stats.append(feature_stats)
    
    # Normalisation des corrélations pour obtenir des poids
    correlation_weights = np.array(correlations)
    correlation_weights = correlation_weights / np.sum(correlation_weights) if np.sum(correlation_weights) > 0 else correlation_weights
    
    # === VISUALISATION PRINCIPALE ===
    print("\n📊 Génération de la visualisation des poids...")
    
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    fig.suptitle('🎯 Analyse des Poids de Contribution des Métriques Biomécaniques', 
                 fontsize=16, fontweight='bold')
    
    # === GRAPHIQUE 1: POIDS PAR CORRÉLATION ===
    ax1 = axes[0, 0]
    bars1 = ax1.bar(range(len(feature_names)), correlation_weights, 
                    color=plt.cm.viridis(correlation_weights / np.max(correlation_weights)))
    ax1.set_title('🔗 Importance par Corrélation avec l\'Expertise', fontweight='bold')
    ax1.set_xlabel('Métriques Biomécaniques')
    ax1.set_ylabel('Poids de Contribution Normalisé')
    ax1.set_xticks(range(len(feature_names)))
    ax1.set_xticklabels(feature_names, rotation=45, ha='right')
    ax1.grid(True, alpha=0.3)
    
    # Ajout des valeurs sur les barres
    for i, v in enumerate(correlation_weights):
        ax1.text(i, v + 0.001, f'{v:.3f}', ha='center', va='bottom', fontsize=8)
    
    # === GRAPHIQUE 2: CLASSEMENT PAR IMPORTANCE ===
    ax2 = axes[0, 1]
    sorted_indices = np.argsort(correlation_weights)[::-1]
    sorted_weights = correlation_weights[sorted_indices]
    sorted_names = [feature_names[i] for i in sorted_indices]
    
    bars2 = ax2.barh(range(len(sorted_names)), sorted_weights, 
                     color=plt.cm.plasma(sorted_weights / np.max(sorted_weights)))
    ax2.set_title('🏆 Classement des Métriques par Importance', fontweight='bold')
    ax2.set_xlabel('Poids de Contribution')
    ax2.set_ylabel('Métriques (ordonnées)')
    ax2.set_yticks(range(len(sorted_names)))
    ax2.set_yticklabels(sorted_names)
    ax2.grid(True, alpha=0.3)
    
    # === GRAPHIQUE 3: RÉPARTITION PAR CATÉGORIE ===
    ax3 = axes[1, 0]
    
    # Regroupement par type de métrique (VRAIES CATÉGORIES DU DATASET)
    categories = {
        'Métriques Principales': ['velocity', 'acceleration', 'jerk'],
        'Jerk Composantes': ['jerk_x', 'jerk_y', 'jerk_z'],
        'Efficacité & Coordination': ['efficiency_index', 'average_separation_distance']
    }
    
    category_weights = {}
    for category, features in categories.items():
        category_weight = sum([correlation_weights[feature_names.index(f)] for f in features if f in feature_names])
        category_weights[category] = category_weight
    
    category_names = list(category_weights.keys())
    category_values = list(category_weights.values())
    
    wedges, texts, autotexts = ax3.pie(category_values, labels=category_names, autopct='%1.1f%%',
                                       colors=plt.cm.Set3(np.arange(len(category_names))))
    ax3.set_title('🔧 Répartition d\'Importance par Type de Métrique', fontweight='bold')
    
    # === GRAPHIQUE 4: ANALYSE DÉTAILLÉE DES MÉTRIQUES ===
    ax4 = axes[1, 1]
    
    # Toutes les métriques (nous en avons 8)
    all_indices = sorted_indices
    all_weights = [correlation_weights[i] for i in all_indices]
    all_names = [feature_names[i] for i in all_indices]
    
    bars4 = ax4.bar(range(len(all_names)), all_weights,
                    color=plt.cm.coolwarm(np.array(all_weights) / np.max(all_weights)))
    ax4.set_title('⭐ Classement Complet des Métriques', fontweight='bold')
    ax4.set_xlabel('Métriques')
    ax4.set_ylabel('Poids d\'Importance')
    ax4.set_xticks(range(len(all_names)))
    ax4.set_xticklabels(all_names, rotation=45, ha='right')
    ax4.grid(True, alpha=0.3)
    
    # Ajout des valeurs
    for i, v in enumerate(all_weights):
        ax4.text(i, v + 0.001, f'{v:.3f}', ha='center', va='bottom', fontsize=9, fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('model_outputs/feature_importance_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    # === RAPPORT TEXTUEL DÉTAILLÉ ===
    print(f"\n📋 === RAPPORT D'IMPORTANCE DES MÉTRIQUES ===")
    print("=" * 70)
    
    print(f"\n🏆 TOP MÉTRIQUES LES PLUS INFLUENTES:")
    print("-" * 50)
    for i in range(min(5, len(sorted_indices))):
        idx = sorted_indices[i]
        name = feature_names[idx]
        weight = correlation_weights[idx]
        print(f"{i+1:2d}. {name:<25} | Poids: {weight:.4f} | Importance: {weight*100:.1f}%")
    
    print(f"\n📊 ANALYSE PAR CATÉGORIE:")
    print("-" * 50)
    for category, weight in sorted(category_weights.items(), key=lambda x: x[1], reverse=True):
        percentage = (weight / sum(category_weights.values())) * 100
        print(f"{category:<15} | Contribution: {weight:.4f} | Part: {percentage:.1f}%")
    
    print(f"\n📈 STATISTIQUES GÉNÉRALES:")
    print("-" * 50)
    print(f"• Métrique la plus influente: {feature_names[sorted_indices[0]]} ({correlation_weights[sorted_indices[0]]:.4f})")
    print(f"• Métrique la moins influente: {feature_names[sorted_indices[-1]]} ({correlation_weights[sorted_indices[-1]]:.4f})")
    print(f"• Écart d'importance: {correlation_weights[sorted_indices[0]] - correlation_weights[sorted_indices[-1]]:.4f}")
    print(f"• Coefficient de variation: {np.std(correlation_weights)/np.mean(correlation_weights):.3f}")
    
    # === RECOMMANDATIONS ===
    print(f"\n💡 RECOMMANDATIONS POUR L'OPTIMISATION:")
    print("-" * 50)
    
    # Identification des métriques critiques (top 50% car nous n'avons que 8 métriques)
    n_critical = max(1, len(feature_names) // 2)
    critical_features = [feature_names[sorted_indices[i]] for i in range(n_critical)]
    
    print(f"🎯 Métriques critiques à surveiller: {', '.join(critical_features)}")
    print(f"📉 Métriques à impact plus limité: {', '.join([feature_names[sorted_indices[i]] for i in range(-2, 0)][::-1])}")
    
    print(f"\n🔍 INTERPRÉTATION DES MÉTRIQUES:")
    print("-" * 50)
    print(f"📊 velocity: Vitesse instantanée du mouvement")
    print(f"⚡ acceleration: Changement de vitesse")
    print(f"🌊 jerk: Fluidité du mouvement (dérivée de l'accélération)")
    print(f"📐 jerk_x/y/z: Composantes directionnelles de la fluidité")
    print(f"🎯 efficiency_index: Ratio vitesse/accélération (efficacité)")
    print(f"🤝 average_separation_distance: Coordination bi-manuelle")
    
    if np.max(correlation_weights) > 0.1:
        print(f"✅ Excellente différenciation des métriques détectée")
    elif np.max(correlation_weights) > 0.05:
        print(f"👍 Bonne répartition de l'importance des métriques")
    else:
        print(f"⚠️ Importance des métriques relativement uniforme - Possible sur-apprentissage")
    
    # === SAUVEGARDE DES RÉSULTATS ===
    importance_results = {
        'feature_importance': {
            'names': feature_names,
            'weights': correlation_weights.tolist(),
            'correlations': [float(c) for c in correlations]
        },
        'category_importance': category_weights,
        'top_features': {
            'names': [feature_names[i] for i in sorted_indices],
            'weights': [float(correlation_weights[i]) for i in sorted_indices]
        },
        'statistics': {
            'max_importance': float(np.max(correlation_weights)),
            'min_importance': float(np.min(correlation_weights)),
            'mean_importance': float(np.mean(correlation_weights)),
            'std_importance': float(np.std(correlation_weights)),
            'coefficient_variation': float(np.std(correlation_weights)/np.mean(correlation_weights))
        }
    }
    
    # Sauvegarde en JSON
    with open('model_outputs/feature_importance_detailed.json', 'w') as f:
        json.dump(importance_results, f, indent=2)
    
    print(f"\n💾 === FICHIERS GÉNÉRÉS ===")
    print(f"📊 Graphique: model_outputs/feature_importance_analysis.png")
    print(f"📄 Données JSON: model_outputs/feature_importance_detailed.json")
    
    print(f"\n✅ === ANALYSE DES POIDS DE CONTRIBUTION TERMINÉE ===")

except Exception as e:
    print(f"❌ Erreur lors de l'analyse d'importance: {e}")
    print(f"🔍 Vérifiez que le modèle et les données sont disponibles")
    import traceback
    traceback.print_exc()
```
